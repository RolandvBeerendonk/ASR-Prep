```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
import datetime
class ASRPREP:
    def __init__(self):
        self.portfolio = pd.DataFrame(columns=['Ticker', 'Sector', 'Asset Class', 'Quantity', 'Purchase Price'])
        self.data = {}
    def add_stock(self, ticker, sector, asset_class, quantity, purchase_price):
        new_entry = {
            'Ticker': ticker,
            'Sector': sector,
            'Asset Class': asset_class,
            'Quantity': quantity,
            'Purchase Price': purchase_price}
        self.portfolio = pd.concat([self.portfolio, pd.DataFrame([new_entry])],ignore_index=True)
    def download_historical_data_portfolio(self,start):
        end = datetime.datetime.today().strftime('%Y-%m-%d')
        tickers=self.portfolio['Ticker']
        for ticker in tickers: 
            self.data[ticker] = yf.download(ticker,start=start,end=end)
    def plot_historical_stock_price(self, tickers=None,combined_graph=False):
        if type(tickers)==str:
            tickers=[tickers]
        %matplotlib inline
        plt.figure(figsize=(12,4))
        close_list=pd.DataFrame()
        if combined_graph is True:
            if tickers is None or len(tickers)==0:
                for ticker in self.data:
                    close=self.data[ticker]['Close']
                    close.plot(label=ticker)
                    plt.title('Historical Closing Prices')
                    plt.xlabel("Date")
                    plt.ylabel("Stock Price[USD]")
                    plt.legend()
                    plt.grid(True)
                    plt.tight_layout()
                    plt.show()
            else:
                for ticker in tickers:
                    close=self.data[ticker]["Close"]
                    close.plot()
                    plt.title('Historical Closing Prices')
                    plt.xlabel("Date")
                    plt.ylabel("Stock Price[USD]")
                    plt.legend()
                    plt.grid(True)
                    plt.tight_layout()
                    plt.show()                 
        if tickers is None or len(tickers)==0:
            for ticker in self.data:
                close=self.data[ticker]['Close']
                close_list=pd.concat([close_list,close],axis=1)
        else:
            for ticker in tickers:
                close=self.data[ticker]["Close"]
                close_list=pd.concat([close_list,close],axis=1)
        close_list.plot(label=ticker)
        plt.title('Historical Closing Prices')
        plt.xlabel("Date")
        plt.ylabel("Stock Price[USD]")
        plt.legend()
        plt.grid(True)
        plt.tight_layout()
        plt.show()
    def current_price(ticker):
        return self.data[ticker]['Close'].iloc[-1]
    def portfolio_value(self):
        self.portfolio['Current Price'] = self.portfolio['Ticker'].apply(current_price)
        self.portfolio['Market Value']= self.portfolioj['Quantity'] * self.portfolio['Current Price']
        total_portfolio_value= self.portfolio['Market Value'].sum()
        return total_portfolio_value
```
