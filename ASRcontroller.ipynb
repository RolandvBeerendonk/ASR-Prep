```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
import datetime
import seaborn as sns
import numpy as np

class ASRController:
    def __init__(self, model, view):
        self.model = model
        self.view = view

    def add_stock(self, ticker, sector, asset_class, quantity, purchase_price):
        self.model.add_stock(ticker, sector, asset_class, quantity, purchase_price)

    def load_data(self, start, end=datetime.datetime.today().strftime('%Y-%m-%d')):
        self.model.download_historical_data_portfolio(start, end)

    def show_current_price(self, ticker):
        return self.model.current_price(ticker)

    def show_portfolio_value(self, by='asset'):
        return self.model.portfolio_value(by=by)

    def show_portfolio_weight(self, by='asset'):
        return self.model.portfolio_weight(by=by)

    def show_cumulative_return(self):
        return_df = self.model.return_price_dataframe()
        self.view.cumulative_return(return_df)

    def show_correlation_plot(self):
        return_df = self.model.return_price_dataframe()
        self.view.correlation_plot(return_df)

    def show_var_plot(self, confidence=0.05):
        return_df = self.model.return_price_dataframe()
        self.view.calculate_plot_var(return_df, confidence=confidence)

    def show_portfolio_summary(self, by='asset'):
        self.model.portfolio['Current Price'] = self.model.portfolio['Ticker'].apply(self.model.current_price)
        self.view.portfolio_summary(self.model.portfolio, by=by)

    def show_portfolio_circle(self, by='asset'):
        self.model.portfolio['Current Price'] = self.model.portfolio['Ticker'].apply(self.model.current_price)
        self.view.portfolio_circle_diagram(self.model.portfolio, by=by)

    def simulate_and_plot_stock_paths(self, n_plot_paths=100):
        simulations = self.model.simulate_stock_paths(number_simulations=n_plot_paths)
        self.view.plot_simulated_paths(simulations, n_plot_paths=n_plot_paths)

    def simulate_and_plot_portfolio_paths(self, mode='sample', n_plot_paths=100):
        portfolio_paths = self.model.simulate_portfolio(number_simulations=n_plot_paths)
        self.view.plot_simulated_portfolio_paths(portfolio_paths, mode=mode, n_plot_paths=n_plot_paths)
    def show_var_and_cvar(self, confidence_level=0.05):
        self.model.print_var_and_cvar(confidence_level=confidence_level)
    def show_historical_prices(self, tickers=None, individual_graph=False):
        close_df = self.model.close_price_dataframe()
        self.view.plot_historical_stock_price(close_df, tickers=tickers, individual_graph=individual_graph)
```
